```python
daily = df_punjab.groupby(['city', 'date']).agg({
    'pm2_5': 'mean',
    'temperature': 'mean',
    'humidity': 'mean',
    'precipitation': 'mean',
    'wind_speed': 'mean',
    'pressure': 'mean',
    'season': 'first',
    'is_weekend': 'first'
}).reset_index()
```
