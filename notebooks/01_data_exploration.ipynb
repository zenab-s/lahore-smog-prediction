```python
daily['threshold'] = daily.groupby('city')['pm2_5'].transform(lambda x : x.quantile(0.8))
daily['hazardous'] = (daily['pm2_5'] > daily['threshold']).astype(int)
```

```python
daily.groupby('city')['threshold'].first()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

daily.groupby('city')['hazardous'].mean().plot(kind='bar', ax=axes[0], title='Hazard Rate by City')
daily.groupby('city')['threshold'].first().plot(kind='bar', ax=axes[1], title='Hazardous Threshold (µg/m³) by City', color='red')

plt.tight_layout()
plt.savefig('../figures/hazard_rate_vs_threshold_by_city.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
hourly_pattern = df_punjab.groupby('hour')['pm2_5'].mean()
hourly_pattern.plot()
plt.title('Average PM2.5 by Hour of Day')
plt.xlabel('Hour')
plt.ylabel('PM2.5 (µg/m³)')
plt.savefig('../figures/hourly_pm25_pattern.png', dpi=150, bbox_inches='tight')
plt.show()
```

After EDA, I found that PM 2.5 showed two daily peaks: one around 6 PM and another around 3 AM (pre-dawn), with a trough around 10 AM. The evening peak likely reflects heavy vehicle emissions during traffic and rush hours. The pre-dawn peak likely reflects a temperature inversion, where cold air near the ground gets trapped beneath a warmer layer above, preventing pollutants from dispersing. The mid-morning trough occurs as the sun warms the ground, breaking the inversion and allowing pollution to disperse.
